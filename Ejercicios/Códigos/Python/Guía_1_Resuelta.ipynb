### Script 1

Desarrolla un script que pida un número al usuario e identifique si el valor ingresado es positivo, negativo o cero.

#### Explicación de mejoras:
- **Validación de tipos:** Se mantiene la validación con `try/except` para prevenir caídas ante ingresos alfabéticos.
- **Claridad:** Salida formateada limpia y explicativa.

```python
# Identificador de números positivos, negativos o cero

while True:
    try:
        numero = float(input("Ingrese un número: "))
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

if numero > 0:
    print(f"El número {numero} es positivo.")
elif numero < 0:
    print(f"El número {numero} es negativo.")
else:
    print("El número es cero.")
```

### Script 4

Desarrolla un script que pida dos números enteros al usuario e indique si el primer número es múltiplo del segundo, utilizando el operador módulo (%).

#### Explicación de mejoras:
- **Corrección del tipo de datos:** El enunciado solicitaba "números enteros", pero el script original utilizaba `float`. Se corrige a `int`.
- **Prevención de excepciones:** Se implementa un filtro de seguridad para impedir que el segundo número (divisor) sea `0`, evitando un error de detención en tiempo de ejecución (`ZeroDivisionError`).

```python
# Identificador de múltiplos con prevención de división por cero

while True:
    try:
        numero = int(input("Ingrese el primer número entero: "))
        while True:
            multiplo = int(input("Ingrese el segundo número entero (distinto de 0): "))
            if multiplo == 0:
                print("El divisor no puede ser cero ya que causará un error matemático. Intente de nuevo.")
            else:
                break
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese números enteros válidos.")

if numero % multiplo == 0:
    print(f"El número {numero} es múltiplo de {multiplo}.")
else:
    print(f"El número {numero} no es múltiplo de {multiplo}.")
```

### Script 13

Desarrolla un script que pida tres números al usuario y determine cuál de ellos es el mayor. Si existen números iguales, el programa también debe indicarlo

#### Explicación de mejoras:
- **Creación de función reutilizable:** El script original repetía exactamente el mismo bloque `while` y `try/except` tres veces para pedir cada número. Se crea la función local `solicitar_numero()` para respetar el diseño DRY.
- **Claridad en igualdades:** Lógica de comparación de igualdades refinada.

```python
# Comparador de tres números (Optimizado con función reutilizable)

print("--- Comparación Avanzada de 3 Números ---")

def solicitar_numero(posicion):
    while True:
        try:
            return float(input(f"Ingrese el {posicion} número: "))
        except ValueError:
            print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

num1 = solicitar_numero("primer")
num2 = solicitar_numero("segundo")
num3 = solicitar_numero("tercer")

mayor = max(num1, num2, num3)
menor = min(num1, num2, num3)

# Verificación de números iguales e impresión informativa
print(f"\n--- Análisis de Resultados ---")
if num1 == num2 == num3:
    print("Información: Los tres números ingresados son exactamente iguales.")
elif num1 == num2 or num1 == num3 or num2 == num3:
    print("Información: Hay dos números que son iguales entre sí.")
else:
    print("Información: Todos los números ingresados son diferentes.")

print(f"Número Mayor: {mayor}")
print(f"Número Menor: {menor}")
```

#### Explicación de diseño:
- **Validación del Teorema de Desigualdad del Triángulo:** Se comprueba matemáticamente si los lados pueden formar un triángulo (`a + b > c`, `a + c > b` y `b + c > a`). Si no se cumple, se informa y descarta, ya que un triángulo inexistente no puede tener clasificación.
- **Clasificación geométrica:** Se verifica la igualdad de sus lados para clasificar en Equilátero, Isósceles o Escaleno.

```python
# Validador y Clasificador de Triángulos

print("--- Clasificador de Triángulos por sus Lados ---")

def solicitar_lado(nombre_lado):
    while True:
        try:
            valor = float(input(f"Ingrese longitud del lado {nombre_lado}: "))
            if valor > 0:
                return valor
            else:
                print("La longitud de un lado debe ser mayor que cero.")
        except ValueError:
            print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

a = solicitar_lado("A")
b = solicitar_lado("B")
c = solicitar_lado("C")

# Validación geométrica de la existencia del triángulo (Desigualdad Triangular)
es_valido = (a + b > c) and (a + c > b) and (b + c > a)

if not es_valido:
    print(f"\nError: Con las longitudes {a}, {b} y {c} NO se puede formar un triángulo físico.")
else:
    # Clasificación geométrica
    if a == b == c:
        tipo = "Equilátero (3 lados iguales)"
    elif a == b or b == c or a == c:
        tipo = "Isósceles (2 lados iguales y 1 distinto)"
    else:
        tipo = "Escaleno (todos sus lados diferentes)"
        
    print(f"\nResultado: El triángulo es {tipo}.")
```

### Script 16 (Nuevo)

Desarrolla un script que pida el peso en kilogramos y la estatura en metros de una persona. Luego, debe calcular el Índice de Masa Corporal (IMC) usando la fórmula:
IMC = peso / (estatura²)
Después, debe mostrar la clasificación correspondiente según la OMS:
- Bajo peso: menor a 18.5
- Normal: desde 18.5 hasta 24.9
- Sobrepeso: 25.0 o más

#### Explicación de diseño:
- **Manejo estricto de rangos reales:** Validación del peso ($\ge 2$ kg) y la estatura ($\ge 0.5$ m) para evitar errores lógicos.
- **Fórmula exacta:** Se calcula y muestra la clasificación estricta de la Organización Mundial de la Salud (OMS).

```python
# Calculador de IMC con estándares de la OMS

print("--- Calculador de Índice de Masa Corporal (IMC) ---")

while True:
    try:
        peso = float(input("Ingrese su peso en kilogramos (ej: 72.5): "))
        if peso > 2.0:
            break
        else:
            print("Por favor, ingrese un peso válido mayor a 2 kg.")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

while True:
    try:
        estatura = float(input("Ingrese su estatura en metros (ej: 1.75): "))
        if estatura > 0.5:
            break
        else:
            print("Por favor, ingrese una estatura válida mayor a 0.5 metros.")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

# Cálculo de IMC
imc = peso / (estatura ** 2)

# Clasificación según la OMS
if imc < 18.5:
    clasificacion = "Bajo Peso"
elif imc < 25.0:
    clasificacion = "Peso Normal"
else:
    clasificacion = "Sobrepeso (se aconseja realizar chequeos preventivos)"

print(f"\n--- Reporte Nutricional ---")
print(f"Peso ingresado: {peso} kg")
print(f"Estatura ingresada: {estatura} m")
print(f"Su IMC es: {imc:.2f}")
print(f"Estado Nutricional (OMS): {clasificacion}")
```

```python
# Calculadora Interactiva con control de errores

print("--- Calculadora Aritmética Interactiva ---")

while True:
    try:
        n1 = float(input("Ingrese el primer número: "))
        n2 = float(input("Ingrese el segundo número: "))
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese valores numéricos.")

operaciones_validas = ["+", "-", "*", "/"]
while True:
    operacion = input("Seleccione la operación (+, -, *, /): ").strip()
    if operacion in operaciones_validas:
        break
    else:
        print("Operador no válido. Ingrese exclusivamente uno de estos caracteres: +, -, *, /")

# Ejecución y control de división por cero
if operacion == "+":
    resultado = n1 + n2
elif operacion == "-":
    resultado = n1 - n2
elif operacion == "*":
    resultado = n1 * n2
else:  # operacion == "/"
    if n2 == 0:
        resultado = "Error: Indefinición matemática (No se puede dividir por cero)"
    else:
        resultado = n1 / n2

print(f"\nResultado de la operación ({n1} {operacion} {n2}): {resultado}")
```

### Script 19 (Nuevo)

Desarrolla un script que pida dos colores primarios entre los siguientes: rojo, amarillo, azul.
El programa debe considerar estas reglas:
- Si ambos colores son iguales, mostrar: “No hay mezcla, es el mismo color”
- Si mezcla rojo + amarillo, mostrar: “naranja”
- Si mezcla rojo + azul, mostrar: “morado”
- Si mezcla amarillo + azul, mostrar: “verde”
- Para cualquier otro caso, mostrar: “Combinación no reconocida”

#### Explicación de diseño:
- **Independencia del orden de entrada:** Se utiliza una colección set (`{color1, color2}`) o lógica con operadores de pertenecia `in` para asegurar que el orden de entrada (ej: "rojo y azul" o "azul y rojo") devuelva el mismo color resultante ("morado").
- **Normalización y validación:** Limpieza y descarte de colores no primarios.

```python
# Mezclador de Colores Primarios con independencia de orden

print("--- Mezclador de Colores Primarios ---")
primarios_validos = ["rojo", "amarillo", "azul"]

def solicitar_color(posicion):
    while True:
        color = input(f"Ingrese el {posicion} color primario (rojo/amarillo/azul): ").strip().lower()
        if color in primarios_validos:
            return color
        else:
            print("Color no reconocido como primario válido. Intente nuevamente.")

c1 = solicitar_color("primer")
c2 = solicitar_color("segundo")

print(f"\nMezclando: {c1} + {c2}...")

if c1 == c2:
    print("Resultado: No hay mezcla, es el mismo color.")
else:
    # Creamos un conjunto para ignorar el orden del ingreso de datos
    mezcla = {c1, c2}
    
    if mezcla == {"rojo", "amarillo"}:
        resultado = "naranja"
    elif mezcla == {"rojo", "azul"}:
        resultado = "morado"
    elif mezcla == {"amarillo", "azul"}:
        resultado = "verde"
    else:
        resultado = "Combinación no reconocida"
        
    print(f"Resultado de la combinación: {resultado.upper()}")
```

### Script 20 (Nuevo)

Desarrolla un script que calcule el peso de un objeto en distintos cuerpos celestes. El programa debe pedir la masa del objeto en kilogramos y luego pedir el cuerpo celeste donde se desea calcular el peso: Tierra, Luna, Marte o Júpiter.
El cálculo debe realizarse con la fórmula:
peso = masa × gravedad
Usa los siguientes valores de gravedad:
- Tierra: 9.8 m/s²
- Luna: 1.62 m/s²
- Marte: 3.71 m/s²
- Júpiter: 24.79 m/s²

El script debe mostrar la masa ingresada, el cuerpo celeste seleccionado y el peso calculado. Si el cuerpo celeste no corresponde a una opción válida, debe mostrar: “Cuerpo celeste no válido”.

#### Explicación de diseño:
- **Física exacta:** Se programa la fórmula exacta $P = m \cdot g$ con control de excepciones y validación de masas positivas.
- **Diccionario de constantes:** Se utiliza un mapeo de gravedad limpio.

```python
# Cálculo de peso en cuerpos celestes

print("--- Cálculo de Peso Gravitacional en el Espacio ---")

gravedades = {
    "tierra": 9.8,
    "luna": 1.62,
    "marte": 3.71,
    "jupiter": 24.79
}

while True:
    try:
        masa = float(input("Ingrese la masa del objeto en kilogramos (ej: 12.8): "))
        if masa > 0:
            break
        else:
            print("La masa del objeto debe ser mayor que cero.")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

cuerpo = input("Ingrese el cuerpo celeste (Tierra / Luna / Marte / Júpiter): ").strip().lower()
# Normalización de caracteres con tildes
if cuerpo == "júpiter":
    cuerpo = "jupiter"

if cuerpo in gravedades:
    g = gravedades[cuerpo]
    peso = masa * g
    print(f"\n--- Ficha de Cálculo Físico ---")
    print(f"Masa ingresada: {masa} kg")
    print(f"Cuerpo celeste seleccionado: {cuerpo.capitalize()}")
    print(f"Aceleración de gravedad g: {g} m/s²")
    print(f"Peso calculado: {peso:.2f} Newtons (N)")
else:
    print("\nCuerpo celeste no válido.")
```
