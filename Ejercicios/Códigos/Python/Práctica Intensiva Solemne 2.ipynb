### Ejercicio 1: Validador de Calificaciones y Rendimiento de Curso

**Enunciado:**
Desarrolla un programa interactivo que solicite al usuario las calificaciones obtenidas por los estudiantes en un examen. Las notas son valores decimales entre **1.0 y 7.0**.

*   El programa debe solicitar notas indefinidamente hasta que el usuario ingrese la nota especial **-1**, la cual actúa como señal de término.
*   **Validación Robusta:** Utilizando bloques `try-except`, debes controlar que la entrada sea efectivamente numérica y esté dentro de la escala permitida (1.0 a 7.0). Si la entrada es inválida (por ejemplo, texto o una nota fuera de rango), se debe mostrar un aviso de error detallado y volver a solicitar el valor sin que la aplicación se caiga.
*   Al finalizar el ingreso de datos, calcula e imprime en pantalla:
    1. El promedio general del curso (redondeado a 1 decimal).
    2. La cantidad total de aprobados (notas mayores o iguales a 4.0).
    3. El porcentaje de aprobación con respecto al total de notas válidas registradas.
    4. La cantidad total de reprobados.

```python
'''
Validador de notas.

- Se reciben notas entre 1.0 hasta 7.0, hasta que se ingrese -1, que es el valor de corte.
- Se debe validar la entrada.
- Se debe mostrar:
    - El promedio general del curso (redondeado a 1 decimal).
    - Cantidad de aprovados (notas >= 4.0).
    - Porcentaje de aprovación con respecto al total de notas ingresadas.
    - Cantidad de reprobados (notas < 4.0).
'''

print("\n --- Validador de notas ---")

#Función de validación

def validar_nota(nota):
    if nota == -1:
        return True
    elif 1.0 <= nota <= 7.0:
        return True
    else:
        return False

notas = []
while True:
    try:
        nota = float(input("Ingrese una nota (1.0 - 7.0) o -1 para finalizar: "))
        if validar_nota(nota):
            if nota == -1:
                break
            notas.append(nota)
        else:
            print("Nota inválida. Por favor, ingrese una nota entre 1.0 y 7.0.")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número.")

if notas:
    promedio = round(sum(notas) / len(notas), 1)
    aprobados = sum(1 for nota in notas if nota >= 4.0)
    reprobados = sum(1 for nota in notas if nota < 4.0)
    porcentaje_aprobacion = (aprobados / len(notas)) * 100

    print(f"\nPromedio general del curso: {promedio}")
    print(f"Cantidad de aprobados: {aprobados}")
    print(f"Porcentaje de aprobación: {porcentaje_aprobacion:.2f}.%")
    print(f"Cantidad de reprobados: {reprobados}")
else:
    print("\nNo se ingresaron notas.")
```

### Ejercicio 2: Gestión de Inventario con Listas Paralelas y Búsqueda Manual

**Enunciado:**
Una distribuidora ferretera requiere gestionar el control de su bodega física de forma estructurada. Escribe un programa modular que haga lo siguiente:

1.  Solicita al usuario ingresar el número de productos diferentes a registrar (debe ser un número entero **mínimo igual a 3**; valida esto bajo un ciclo interactivo).
2.  Mediante listas paralelas (`productos`, `stocks`, `precios`), solicita de forma secuencial:
    *   El nombre del producto (`str`).
    *   La cantidad en stock disponible (un entero no negativo).
    *   El precio unitario de venta (un entero positivo).
3.  Al concluir el registro interactivo:
    *   Imprime en consola una tabla elegante y alineada detallando cada producto con sus respectivos valores.
    *   Calcula y muestra el **valor totalizado en dinero de todo el inventario** (la suma acumulada de multiplicar el stock por el precio unitario en cada artículo).
    *   **Algoritmo Extremo Manual:** Busca de forma secuencial y muestra en pantalla el producto que tiene el **menor stock registrado** (indicando su nombre y cantidad). **Queda prohibido utilizar la función embebida `min()` o `.index()`; debes diseñar el recorrido lineal manual.**

```python
'''
Gestión de listas y busqueda manual

Se pide:
    - Cantidad de productos distintos a ingresar (minimo 3)

Según la cantidad de productos, se hace una lista de:
    - Nombre del producto
    -Cantidad de stock
    -Precio unitario

Se debe mostrar:
    - Una tabla con cada producto y su valor.
    - El valor total del inventario (stock * precio unitario).

Restricción:
    - Busca de forma secuencial y muestra en pantalla el producto que tiene el **menor stock registrado** 
    (indicando su nombre y cantidad). **Queda prohibido utilizar la función embebida `min()` o `.index()`; 
    debes diseñar el recorrido lineal manual.**
'''

print("\n --- Registro de inventario ---")

# Función de validación
def validar_producto(mensaje, minimo):
    while True:
        try:
            valor = int(input(mensaje))
            if valor >= minimo:
                return valor
            else:
                print(f"Debe ingresar un número entero mayor o igual a {minimo}.")
        except ValueError:
            print("Entrada no válida. Por favor, ingrese un número entero.")


# Solicitar cantidad de productos
cantidad_productos = validar_producto("Ingrese la cantidad de productos distintos a registrar (mínimo 3): ", 3)

productos = []
stock = []
precios = []

for i in range(cantidad_productos):
    nombre = input(f"\nIngrese el nombre del producto {i + 1}: ")
    productos.append(nombre)

    cantidad_stock = validar_producto(f"Ingrese la cantidad de stock para {nombre}: ", 0)
    stock.append(cantidad_stock)

    precio_unitario = validar_producto(f"Ingrese el precio unitario para {nombre}: ", 0)
    precios.append(precio_unitario)

# Mostrar tabla de productos
print("\n --- Inventario de productos ---")
print(f"{'Producto':<20} {'Stock':<10} {'Precio Unitario':<15} {'Valor Total':<15}")
print("-" * 60)
valor_total_inventario = 0
for i in range(cantidad_productos):
    valor_total = stock[i] * precios[i]
    valor_total_inventario += valor_total
    print(f"{productos[i]:<20} {stock[i]:<10} {precios[i]:<15} {valor_total:<15}")

print("-" * 60)
print(f"{'Valor Total del Inventario':<45} {valor_total_inventario:<15}")

# Búsqueda del producto con menor stock
menor_stock = stock[0]
producto_menor_stock = productos[0]
for i in range(0, cantidad_productos):
    if stock[i] < menor_stock:
        menor_stock = stock[i]
        producto_menor_stock = productos[i]
print(f"\nProducto con menor stock: {producto_menor_stock} (Cantidad: {menor_stock})")
```

### Ejercicio 3: Calculadora de Encomiendas con Tarifas Diferenciadas por Destino

**Enunciado:**
Diseña una aplicación para liquidar el despacho de paquetes postales con recargo por distancia. 

1.  Define una función modular llamada `calcular_tarifa(peso, destino)` que retorne el costo de un despacho individual. La función recibe el peso en kilogramos (`float`) y el destino como una cadena de un solo carácter:
    *   **"U" (Urbano):** Cobrar $3.000 por kg para los primeros 5 kg, y $2.000 por cada kg adicional sobre ese límite.
    *   **"R" (Rural):** Cobrar $4.500 por kg para los primeros 5 kg, y $3.500 por cada kg adicional.
    *   **"E" (Extranjero/Internacional):** Cobrar $12.000 fijos por cada kilogramo, sin importar el peso.
2.  El programa principal debe solicitar de forma continua el registro de múltiples envíos. Para cada paquete se captura el peso (que debe ser mayor a 0 kg) y el código de destino (se debe validar que solo acepte "U", "R" o "E", sin importar si se ingresa en mayúsculas o minúsculas).
3.  El ingreso de paquetes finaliza cuando el usuario escribe una tarifa de peso igual a **0**.
4.  Al concluir, se debe imprimir en pantalla la cantidad de envíos procesados y el cobro acumulado total de todas las liquidaciones despachadas.

```python
'''
Calculadora de tarifa

Valores de entrada:
    -Peso del paquete (kg) (mayor que 0)
    - Código de zona ("U" (urbano), "R" (Rural) o "E" (Extranjero/internacional)). Cada zona tiene un valor:
        - U: $3000 para los primeros 5 kg y $2000 por cada kg adicional.
        - R: $4500 para los primeros 5 kg y $3500 por cada kg adicional.
        - E: $12000 por cada kilogramo, sin importar el peso.

Se debe generar una función calolar_tarifa(peso, destino). El programa debe resgistar envios hasta que se ingrese peso=0
Para el código de zona, no importa si se ingresa en mayúscula o minúscula, pero debe ser una de las opciones válidas.

Se debe mostrar:
    - Cantidad de envios
    - Cobro acumulado
'''
print("\n --- Calculadora de tarifa ---")

# Funciones de validación
def validar_peso(mensaje):
    while True:
        try:
            valor = float(input(mensaje))
            if valor >= 0:
                return valor
            else:
                print("Debe ingresar un número mayor o igual a 0.")
        except ValueError:
            print("Entrada no válida. Por favor, ingrese un número.")

def validar_destino(mensaje):
    while True:
        valor = input(mensaje).upper()
        if valor in ["U", "R", "E"]:
            return valor
        else:
            print("Destino no válido. Ingrese 'U' (Urbano), 'R' (Rural) o 'E' (Extranjero).")

# Función para calcular tarifa
def calcular_tarifa(peso, destino):
    if destino == "U":
        if peso <= 5:
            return 3000
        else:
            return 3000 + (peso - 5) * 2000
    elif destino == "R":
        if peso <= 5:
            return 4500
        else:
            return 4500 + (peso - 5) * 3500
    elif destino == "E":
        return peso * 12000

cantidad_envios = 0
cobro_acumulado = 0

while True:
    peso = validar_peso("Ingrese el peso del paquete en kg (0 para finalizar): ")
    if peso == 0:
        break
    destino = validar_destino("Ingrese el código de zona (U, R, E): ")
    
    tarifa = calcular_tarifa(peso, destino)
    cobro_acumulado += tarifa
    cantidad_envios += 1

print(f"\nCantidad de envíos: {cantidad_envios}")
print(f"Cobro acumulado: ${cobro_acumulado}")
```
