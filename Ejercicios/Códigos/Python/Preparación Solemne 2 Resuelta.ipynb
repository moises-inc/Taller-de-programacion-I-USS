## 1. Enunciado del Problema y Reglas de Negocio

Una pequeña empresa de construcción necesita calcular la planilla de sueldos mensual de sus trabajadores. El pago se calcula según las horas trabajadas y una tarifa por hora individual para cada trabajador.

### Reglas de cálculo:
*   Las primeras **45 horas mensuales** se pagan a tarifa normal.
*   Las horas que superen las 45 se pagan con un **recargo del 50%** (tarifa × 1.5).
*   `Pago Total = Pago Base + Pago Extra`

### Restricciones del Diseño:
*   Uso estricto de **listas paralelas**.
*   Definir una **función especializada** para cada bloque de cálculo.
*   **Validación robusta** de las horas y las tarifas como valores positivos.

## 2. Bloque 1 - Constantes del Programa

Definición de las constantes que guían las reglas comerciales del caso de estudio.

```python
# Constantes del programa: defínelas al inicio, fuera de cualquier función
HORAS_BASE = 45        # Límite de horas sin recargo comercial
RECARGO_EXTRA = 1.5     # Factor de recargo para las horas extra (50% más)
SEPARADOR = "=" * 68   # Línea divisoria estética para los reportes de consola
```

## 3. Bloque 2 - Función Auxiliar de Validación

Función modular para capturar valores numéricos positivos, evitando duplicidades de try-except.

```python
def ingresar_numero(mensaje, minimo=0.01):
    """
    Solicita un número decimal mayor o igual que "minimo".
    Repite la pregunta de forma interactiva hasta que el dato sea válido.
    Retorna el valor ingresado.
    """
    while True:
        try:
            valor = float(input(mensaje))
            if valor < minimo:
                print(f"   [Error] Ingrese un valor mayor o igual a {minimo}.")
            else:
                return valor # Retorno del valor validado y salida del bucle
        except ValueError:
            print("   [Error] Ingrese un número válido (use punto decimal).")
```

## 4. Bloque 3 - Ingreso de Datos con Listas Paralelas

Lectura de la nómina de trabajadores aplicando validación de cantidad e ingresos por teclado.

```python
def ingresar_trabajadores():
    """
    Pide los datos de n trabajadores (n >= 2).
    Retorna tres listas paralelas: nombres, horas, tarifas.
    """
    # Captura y validación de la cantidad de trabajadores
    while True:
        try:
            n = int(input(" Cantidad de trabajadores (min 2): "))
            if n >= 2:
                break
            print("   [Error] Deben ser al menos 2 trabajadores.")
        except ValueError:
            print("   [Error] Ingrese un número entero válido.")
            
    nombres = []
    horas = []
    tarifas = []
    
    # Ciclo de lectura individual interactiva
    for i in range(n):
        print(f"\n Trabajador {i+1} de {n}")
        nombre = input(" Nombre : ").strip()
        while not nombre:
            nombre = input("   [Error] El nombre es obligatorio. Nombre : ").strip()
            
        h = ingresar_numero(" Horas del mes : ", 1)
        t = ingresar_numero(" Tarifa ($/hora) : ", 1)
        
        # Inserción paralela en las listas
        nombres.append(nombre)
        horas.append(h)
        tarifas.append(t)
        
    return nombres, horas, tarifas
```

## 5. Bloque 4 - Cálculo del Pago de un Trabajador

Lógica comercial para la liquidación mensual de horas e incremento de recargo de horas extra.

```python
def calcular_pago(horas, tarifa):
    """
    Recibe las horas trabajadas y la tarifa por hora.
    Retorna una tupla: (pago_base, pago_extra, pago_total).
    """
    if horas <= HORAS_BASE:
        # Caso sin horas extras
        horas_normales = horas
        horas_extra = 0
    else:
        # Caso con horas extras
        horas_normales = HORAS_BASE
        horas_extra = horas - HORAS_BASE
        
    # Aritmética de liquidación
    pago_base = horas_normales * tarifa
    pago_extra = horas_extra * tarifa * RECARGO_EXTRA
    pago_total = pago_base + pago_extra
    
    return pago_base, pago_extra, pago_total
```

## 6. Bloque 5 - Buscar el Mayor y Menor Sueldo

Algoritmo clásico de recorrido secuencial lineal ($O(N)$) para determinar extremos de liquidación.

```python
def encontrar_extremo(nombres, pagos, buscar_maximo):
    """
    Busca el trabajador con el mayor o menor sueldo según el booleano buscar_maximo.
    Retorna una tupla: (nombre, pago) del trabajador encontrado.
    """
    idx_extremo = 0 # Asunción inicial en el índice 0
    
    # Comparación lineal de un paso
    for i in range(1, len(pagos)):
        if buscar_maximo and pagos[i] > pagos[idx_extremo]:
            idx_extremo = i
        if not buscar_maximo and pagos[i] < pagos[idx_extremo]:
            idx_extremo = i
            
    return nombres[idx_extremo], pagos[idx_extremo]
```

## 7. Bloque 6 - Mostrar Resultados

Presentación gráfica tabular en consola alineando columnas y desplegando agregaciones globales.

```python
def mostrar_resultados(nombres, horas, pagos_base, pagos_extra, pagos_totales):
    """
    Imprime la tabla completa de la planilla mensual de sueldos y consolida totales.
    Recibe las listas ya calculadas para impresión.
    """
    print(f"\n{SEPARADOR}")
    print(" PLANILLA DE SUELDOS MENSUAL")
    print(SEPARADOR)
    
    # Encabezado formateado por columnas
    print(f" {'NOMBRE':<20} {'H.NORM':>6} {'H.EXT':>6} {'PAGO BASE':>10} {'P.EXTRA':>9} {'TOTAL':>10}")
    print(" " + "-" * 64)
    
    # Detalle individual interactivo por trabajador
    for i in range(len(nombres)):
        h_norm = min(horas[i], HORAS_BASE)
        h_extra = max(0, horas[i] - HORAS_BASE)
        print(f" {nombres[i]:<20} {h_norm:>6.0f} {h_extra:>6.0f} "
              f"{pagos_base[i]:>10,.0f} {pagos_extra[i]:>9,.0f} {pagos_totales[i]:>10,.0f}")
        
    # Agregación general sumando la lista de sueldos totales
    total_planilla = sum(pagos_totales)
    print(f"\n TOTAL PLANILLA : ${total_planilla:,.0f}")
    
    # Determinación e informe de trabajadores extremos
    nombre_max, pago_max = encontrar_extremo(nombres, pagos_totales, True)
    nombre_min, pago_min = encontrar_extremo(nombres, pagos_totales, False)
    
    print(f" Mayor sueldo : {nombre_max} --> ${pago_max:,.0f}")
    print(f" Menor sueldo : {nombre_min} --> ${pago_min:,.0f}")
    print(SEPARADOR)
```

## 8. Bloque 7 - Función Main y Punto de Entrada

Orquestación secuencial del flujo comercial uniendo el ingreso, cómputo y despliegue.

```python
def main():
    """
    Orquestador principal que llama a los bloques modulares en orden lógico.
    """
    # Paso 1: Registro e ingreso de datos
    nombres, horas, tarifas = ingresar_trabajadores()
    
    # Inicialización de listas colectoras para resultados
    pagos_base = []
    pagos_extra = []
    pagos_totales = []
    
    # Paso 2: Ciclo de cómputos paralelos
    for i in range(len(nombres)):
        base, extra, total = calcular_pago(horas[i], tarifas[i])
        pagos_base.append(base)
        pagos_extra.append(extra)
        pagos_totales.append(total)
        
    # Paso 3: Despliegue del informe de liquidaciones
    mostrar_resultados(nombres, horas, pagos_base, pagos_extra, pagos_totales)

if __name__ == "__main__":
    main()
```
