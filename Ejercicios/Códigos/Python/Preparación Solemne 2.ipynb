## 1. Enunciado del Problema (Caso G)

Una pequeña empresa de construcción necesita calcular la planilla de sueldos mensual de sus trabajadores. El pago se calcula según las horas trabajadas y una tarifa por hora individual para cada trabajador.

### Datos de entrada:
*   Número de trabajadores (mínimo 2).
*   Para cada trabajador: nombre, horas trabajadas en el mes y tarifa por hora ($).

### Reglas de cálculo:
*   Las primeras **45 horas mensuales** se pagan a tarifa normal.
*   Las horas que superen las 45 se pagan con un **recargo del 50%** (es decir, tarifa × 1.5).
*   `Pago Total = Pago Base + Pago Extra`

### Resultados requeridos:
*   **Tabla formateada** detallando: Nombre, Horas Normales, Horas Extra, Pago Base, Pago Extra y Pago Total.
*   **Total de la planilla** (la suma acumulada de todos los pagos totales).
*   **Identificación de extremos:** Nombre y sueldo del trabajador con el mayor sueldo, y del trabajador con el menor sueldo.

### ⚠️ Restricciones del Diseño:
1.  **Listas Paralelas:** Se debe usar una lista independiente por cada campo (`nombres`, `horas`, `tarifas`, `pagos_base`, `pagos_extra`, `pagos_totales`).
2.  **Modularidad:** Debes definir una función especializada para cada bloque de cálculo.
3.  **Validación Robusta:** Asegura que las horas y las tarifas ingresadas sean valores estrictamente positivos mediante captura con manejo de excepciones (`try-except`).

```python
''' 
Plantilla de sueldo

Datos de entrada:
- Numero de trabajadores (mínimo 2)
- Nombre del trabajador
- Horas trabajadas en el mes
- Valor de la hora de trabajo

Cálculo:
- Las primeras **45 horas mensuales** se pagan a tarifa normal.
- Las horas que superen las 45 se pagan con un **recargo del 50%** (es decir, tarifa * 1.5).
- Pago Total = Pago Base + Pago Extra

Se requiere:
- Lista de cada dato.
- Uso de funciones.
- Validación.

Se debe mostrar: 
*   **Tabla formateada** detallando: Nombre, Horas Normales, Horas Extra, Pago Base, Pago Extra y Pago Total.
*   **Total de la planilla** (la suma acumulada de todos los pagos totales).
*   **Identificación de extremos:** Nombre y sueldo del trabajador con el mayor sueldo, y del trabajador con el menor sueldo.
'''

print("\nBienvenido al programa de cálculo de sueldos")

#Función de validación
def validar_entero(mensaje, minimo):
    while True:
        try:
            valor = int(input(mensaje))
            if valor < minimo:
                print(f"Por favor, ingrese un número mayor o igual a {minimo}.")
            else:
                return valor
        except ValueError:
            print("Entrada no válida. Por favor, ingrese un número entero.")

#Función para calcular el sueldo
def calcular_sueldo(horas_trabajadas, valor_hora):
    if horas_trabajadas <= 45:
        pago_total = horas_trabajadas * valor_hora
    else:
        pago_base = 45 * valor_hora
        horas_extra = horas_trabajadas - 45
        pago_extra = horas_extra * valor_hora * 1.5
        pago_total = pago_base + pago_extra
    return pago_total

# Solicitar número de trabajadores
num_trabajadores = validar_entero("Ingrese el número de trabajadores (mínimo 2): ", 2)

# Listas para almacenar datos
nombres = []
horas_trabajadas = []
valor_hora = []
sueldos = []

# Recolección de datos
for i in range(num_trabajadores):
    nombre = input(f"Ingrese el nombre del trabajador {i + 1}: ")
    horas = validar_entero(f"Ingrese las horas trabajadas por {nombre} en el mes: ", 0)
    valor = validar_entero(f"Ingrese el valor de la hora de trabajo para {nombre}: ", 0)

    nombres.append(nombre)
    horas_trabajadas.append(horas)
    valor_hora.append(valor)

    sueldo = calcular_sueldo(horas, valor)
    sueldos.append(sueldo)

#Función para mostrar la tabla formateada
def mostrar_tabla(nombres, horas_trabajadas, valor_hora, sueldos):
    print("\nTabla de Sueldos:")
    print(f"{'Nombre':<20} {'Horas Normales':<15} {'Horas Extra':<12} {'Pago Base':<10} {'Pago Extra':<10} {'Pago Total':<10}")
    print("-" * 80)
    for i in range(len(nombres)):
        horas_normales = min(horas_trabajadas[i], 45)
        horas_extra = max(0, horas_trabajadas[i] - 45)
        pago_base = horas_normales * valor_hora[i]
        pago_extra = horas_extra * valor_hora[i] * 1.5
        pago_total = sueldos[i]
        print(f"{nombres[i]:<20} {horas_normales:<15} {horas_extra:<12} {pago_base:<10.2f} {pago_extra:<10.2f} {pago_total:<10.2f}")

#función para identificar extremos
def identificar_extremos(nombres, sueldos):
    max_sueldo = max(sueldos)
    min_sueldo = min(sueldos)
    trabajador_max = nombres[sueldos.index(max_sueldo)]
    trabajador_min = nombres[sueldos.index(min_sueldo)]
    print(f"\nTrabajador con el mayor sueldo: {trabajador_max} con un sueldo de {max_sueldo:.2f}")
    print(f"Trabajador con el menor sueldo: {trabajador_min} con un sueldo de {min_sueldo:.2f}")

# Mostrar la tabla
mostrar_tabla(nombres, horas_trabajadas, valor_hora, sueldos)

# Calcular total de la planilla
total_planilla = sum(sueldos)
print(f"\nTotal de la planilla: {total_planilla:.2f}")

# Identificar extremos
identificar_extremos(nombres, sueldos)
```
