### Script 1: Identificador de números positivos, negativos o cero

**Enunciado:** Desarrolla un script que pida un número al usuario e identifique si el valor ingresado es positivo, negativo o cero.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Se debe solicitar un número y clasificarlo comparándolo contra cero (mayor que cero, menor que cero o exactamente cero).
- **Lógica de la solución:**
1. Capturar el número por teclado.
2. Validar con un bloque `try-except` que la entrada sea de tipo numérico decimal (`float`).
3. Utilizar bifurcaciones condicionales `if-elif-else` para clasificar y mostrar el resultado.

```python
# Script 1: Identificador de números positivos, negativos o cero

while True:
    try:
        numero = float(input("Ingrese un número: "))
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número decimal o entero.")

if numero > 0:
    print(f"El número {numero} es positivo.")
elif numero < 0:
    print(f"El número {numero} es negativo.")
else:
    print("El número es cero.")
```

### Script 2: Identificador de par o impar

**Enunciado:** Desarrolla un script que pida un número entero al usuario e identifique si es par o impar utilizando el operador módulo (%).

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Determinar la paridad de un número entero. Los números impares dejan un resto de 1 al dividirse por 2, y los pares un resto de 0.
- **Lógica de la solución:**
1. Capturar un número forzando el tipo `int` (entero).
2. Evaluar el residuo de la división por 2 usando el operador módulo `%`.
3. Si `numero % 2 == 0` es verdadero es par; de lo contrario es impar.

```python
# Script 2: Identificador de números pares e impares

while True:
    try:
        numero = int(input("Ingrese un número entero: "))
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número entero.")

if numero % 2 == 0:
    print(f"El número {numero} es par.")
else:
    print(f"El número {numero} es impar.")
```

### Script 3: Comparador de dos números

**Enunciado:** Desarrolla un script que pida dos números al usuario y determine cuál de ellos es mayor, o bien indique si ambos son iguales.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Comparar dos valores numéricos reales e informar de forma explícita cuál es el mayor, o bien si existe coincidencia total.
- **Lógica de la solución:**
1. Solicitar y capturar secuencialmente dos números de tipo `float`.
2. Comparar usando `>` y `<` para determinar cuál es mayor.
3. Si ambas condiciones fallan, por exclusión matemática, los números son iguales.

```python
# Script 3: Comparación de mayor, menor o igual entre dos números

while True:
    try:
        numero1 = float(input("Ingrese el primer número: "))
        numero2 = float(input("Ingrese el segundo número: "))
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese valores numéricos.")

if numero1 > numero2:
    print(f"El número {numero1} es mayor que {numero2}.")
elif numero1 < numero2:
    print(f"El número {numero1} es menor que {numero2}.")
else:
    print(f"Ambos números son iguales ({numero1}).")
```

### Script 4: Identificador de múltiplos

**Enunciado:** Desarrolla un script que pida dos números enteros al usuario e indique si el primer número es múltiplo del segundo, utilizando el operador módulo (%).

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Determinar si el primer número entero es divisible exactamente por el segundo, previniendo fallos críticos de división por cero.
- **Lógica de la solución:**
1. Capturar dos números enteros (`int`).
2. Aplicar un bucle de validación secundaria sobre el segundo número para garantizar que sea diferente de cero.
3. Evaluar la divisibilidad con `numero % multiplo == 0`.

```python
# Script 4: Identificador de múltiplos con control de división por cero

while True:
    try:
        numero = int(input("Ingrese el primer número entero: "))
        while True:
            multiplo = int(input("Ingrese el segundo número entero (distinto de 0): "))
            if multiplo == 0:
                print("El divisor no puede ser cero ya que causará un error matemático. Intente de nuevo.")
            else:
                break
        break
    except ValueError:
        print("Entrada no válida. Por favor, ingrese números enteros válidos.")

if numero % multiplo == 0:
    print(f"El número {numero} es múltiplo de {multiplo}.")
else:
    print(f"El número {numero} no es múltiplo de {multiplo}.")
```

### Script 8: Calculador de edad actual

**Enunciado:** Desarrolla un script que pida el nombre de un usuario y su año de nacimiento, y luego muestre un mensaje indicando el nombre de la persona y la edad que tiene actualmente.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Calcular dinámicamente la edad del estudiante a partir de su año de nacimiento, empleando el reloj del sistema a través del módulo standard `datetime`.
- **Lógica de la solución:**
1. Solicitar el nombre evitando registros en blanco.
2. Capturar el año de nacimiento como entero (`int`).
3. Calcular la edad restando al año del sistema `datetime.datetime.now().year` el año ingresado.

```python
# Script 8: Calculador dinámico de edad actual

import datetime

ano_actual = datetime.datetime.now().year

nombre = input("Ingrese su nombre: ").strip()
while not nombre:
    nombre = input("El nombre no puede estar vacío. Ingrese su nombre: ").strip()

while True:
    try:
        ano_nacimiento = int(input(f"Ingrese su año de nacimiento (entre 1900 y {ano_actual}): "))
        if 1900 <= ano_nacimiento <= ano_actual:
            break
        else:
            print(f"El año debe ser válido y de este siglo o el anterior (1900 a {ano_actual}).")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un año en formato entero.")

edad = ano_actual - ano_nacimiento
print(f"Hola {nombre}, según tu año de nacimiento, tu edad actual es: {edad} años.")
```

### Script 9: Cálculo de cobro de estacionamiento (DRY)

**Enunciado:** Desarrolla un script que pida la patente de un vehículo y la cantidad de horas enteras estacionado. El estacionamiento cobra según el siguiente criterio:
- Hasta 2 horas, cobra $2.000 en total.
- Más de 2 horas y hasta 5 horas, cobra $3.500 en total.
- Más de 5 horas, cobra $5.000 en total.
El script debe mostrar la patente y el valor a pagar.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Calcular la tarifa final de arriendo de plaza basándose en las horas enteras consumidas, respetando estrictamente DRY.
- **Lógica de la solución:**
1. Solicitar y normalizar la patente.
2. Solicitar y validar la cantidad de horas enteras (tipo entero positivo >= 0).
3. Determinar el costo final de manera condicionada y realizar una única salida formateada al final del algoritmo.

```python
# Script 9: Cobro de estacionamiento (Optimizado con DRY)

patente = input("Ingrese la patente del vehículo: ").strip().upper()
while not patente:
    patente = input("La patente es obligatoria. Ingrese la patente: ").strip().upper()

while True:
    try:
        horas = int(input("Ingrese la cantidad de horas enteras estacionado: "))
        if horas >= 0:
            break
        else:
            print("Las horas no pueden ser negativas.")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número entero.")

# Determinación de costo
if horas <= 2:
    costo = 2000
elif horas <= 5:
    costo = 3500
else:
    costo = 5000

# Salida única formateada (DRY)
print(f"\nVehículo Patente: {patente} | Horas estacionado: {horas} hr(s) | Total a pagar: ${costo:,} CLP")
```

### Script 10: Cálculo de forma de pago y recargos/descuentos

**Enunciado:** Desarrolla un script que considere un producto con precio base de $80.000. Luego, debe pedir al usuario la forma de pago: efectivo, débito o crédito. Aplica las siguientes reglas:
- Si paga con efectivo, se realiza un descuento de $5.000.
- Si paga con crédito, se aplica un recargo de $3.000.
- Si paga con débito, el precio no cambia.
El script debe mostrar el precio final a pagar.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Calcular la variación monetaria sobre una compra base según el medio de pago, normalizando la entrada del usuario.
- **Lógica de la solución:**
1. Definir el precio constante ($80,000).
2. Solicitar el método de pago limpiando espacios y tildes mediante `.strip().lower()`.
3. Modificar el monto total sumando o restando el recargo o descuento respectivo, imprimiendo un resumen consolidado final.

```python
# Script 10: Venta en Punto de Venta con variaciones de pago (DRY)

precio_base = 80000

while True:
    forma_pago = input("Ingrese forma de pago (efectivo / débito / crédito): ").strip().lower()
    if forma_pago == "crédito":
        forma_pago = "credito"
    elif forma_pago == "débito":
        forma_pago = "debito"
        
    if forma_pago in ["efectivo", "credito", "debito"]:
        break
    else:
        print("Método de pago inválido. Intente nuevamente.")

if forma_pago == "efectivo":
    variacion = -5000
    detalle = "Descuento por Efectivo: -$5.000"
elif forma_pago == "credito":
    variacion = 3000
    detalle = "Recargo por Tarjeta de Crédito: +$3.000"
else:
    variacion = 0
    detalle = "Sin descuentos ni recargos en Débito"

precio_final = precio_base + variacion

print(f"\n--- Resumen de Venta ---")
print(f"Precio Base: ${precio_base:,}")
print(f"Detalle: {detalle}")
print(f"Total Neto a Pagar: ${precio_final:,} CLP")
```

### Script 11: Control de notas para 30 estudiantes con ciclos

**Enunciado:** Desarrolla un script que solicite las notas finales de 30 estudiantes, en escala de 1.0 a 7.0. Para cada estudiante, debe indicar si está aprobado o reprobado. Al final, debe mostrar:
• Cantidad de aprobados
• Cantidad de reprobados
• Promedio general del curso

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Capturar iterativamente 30 notas académicas chilenas válidas. Mostrar la condición individual al instante y consolidar al final contadores generales y la media general del grupo.
- **Lógica de la solución:**
1. Definir contadores `aprobados = 0`, `reprobados = 0` y un acumulador decimal `suma_notas = 0.0`.
2. Iterar 30 veces con un ciclo `for i in range(30)`.
3. En cada ciclo, anidar un bucle `while True` interactivo que valide la captura de la nota flotante (1.0 a 7.0).
4. Evaluar la nota: si `nota >= 4.0` sumar 1 a `aprobados`, sino sumar 1 a `reprobados`.
5. Sumar la nota a `suma_notas`. Calcular el promedio final dividiendo por 30.

```python
# Script 11: Control de notas para 30 estudiantes con ciclos

print("--- Registro de Notas (30 Estudiantes) ---")
aprobados = 0
reprobados = 0
suma_notas = 0.0

for i in range(30):
    while True:
        try:
            nota = float(input(f"Ingrese la nota del estudiante {i+1} (1.0 a 7.0): "))
            if 1.0 <= nota <= 7.0:
                break
            print("Error: La nota debe estar en el rango de 1.0 a 7.0.")
        except ValueError:
            print("Error: Ingrese un número decimal válido.")
    
    suma_notas += nota
    if nota >= 4.0:
        aprobados += 1
        print(f"Estudiante {i+1} Aprobado con nota {nota:.1f}")
    else:
        reprobados += 1
        print(f"Estudiante {i+1} Reprobado con nota {nota:.1f}")

promedio = suma_notas / 30
print("\n--- Resumen del Curso ---")
print(f"Cantidad de aprobados: {aprobados}")
print(f"Cantidad de reprobados: {reprobados}")
print(f"Promedio general del curso: {promedio:.2f}")
```

### Script 12: Caja registradora fija de 25 clientes

**Enunciado:** Desarrolla un script que procese la compra de 25 clientes en una tienda. Cada cliente compra un producto con precio base de $80.000. El programa debe pedir la forma de pago: efectivo, débito o crédito. Aplica las siguientes reglas:
• Si paga con efectivo, se realiza un descuento de $5.000.
• Si paga con crédito, se aplica un recargo de $3.000.
• Si paga con débito, el precio no cambia.
Al final, debe mostrar: Total recaudado, cantidad de pagos en efectivo, débito y crédito.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Procesar en bucle de 25 iteraciones el cobro e incremento acumulado en la caja de una tienda comercial, aplicando reglas dinámicas de pago.
- **Lógica de la solución:**
1. Definir constantes e inicializar acumulador de dinero y contadores de medios de pago.
2. Iterar 25 veces preguntando el medio de pago normalizado.
3. Clasificar con condiciones para ajustar el descuento, recargo e incrementar el contador respectivo.
4. Mostrar al cliente su total y acumular la cifra a la caja final del día.

```python
# Script 12: Procesamiento de ventas para 25 clientes

print("--- Punto de Venta (25 Clientes) ---")
precio_base = 80000
total_recaudado = 0
efectivo_cnt = 0
debito_cnt = 0
credito_cnt = 0

for i in range(25):
    print(f"\n--- Cliente {i+1} ---")
    while True:
        forma_pago = input("Ingrese forma de pago (efectivo / debito / credito): ").strip().lower()
        if forma_pago == "efectivo":
            descuento = 5000
            recargo = 0
            efectivo_cnt += 1
            break
        elif forma_pago in ["credito", "crédito"]:
            descuento = 0
            recargo = 3000
            credito_cnt += 1
            break
        elif forma_pago in ["debito", "débito"]:
            descuento = 0
            recargo = 0
            debito_cnt += 1
            break
        else:
            print("Método de pago no válido. Reintente.")

    precio_final = precio_base - descuento + recargo
    total_recaudado += precio_final
    print(f"Precio final Cliente {i+1}: ${precio_final:,} CLP")

print("\n--- Reporte de Cierre de Caja ---")
print(f"Total recaudado en el día: ${total_recaudado:,} CLP")
print(f"Pagos en efectivo: {efectivo_cnt}")
print(f"Pagos con débito: {debito_cnt}")
print(f"Pagos con crédito: {credito_cnt}")
```

### Script 13: Control de cobros para 20 vehículos

**Enunciado:** Desarrolla un script que permita registrar la entrada de 20 vehículos a un estacionamiento. Para cada vehículo, debe pedir Patente y Horas enteras estacionado. Aplica las tarifas:
• Hasta 2 horas: $2.000
• Más de 2 horas y hasta 5 horas: $3.500
• Más de 5 horas: $5.000
Al final, debe mostrar: Total recaudado, cantidad de vehículos en cada tramo de cobro.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Registrar iterativamente patentes y horas de 20 vehículos para calcular el arriendo y tramos del total recaudado.
- **Lógica de la solución:**
1. Definir variables acumuladoras de dinero y contadores por tramos tarifarios.
2. Utilizar bucle de 20 repeticiones. Solicitar patente y validar que no esté en blanco.
3. Solicitar horas validando enteros no negativos. Aplicar condiciones de tramos y sumar al total.

```python
# Script 13: Control de estacionamiento (20 Vehículos)

print("--- Control de Estacionamiento (20 Vehículos) ---")
total_recaudado = 0
tramo_1_cnt = 0  # Hasta 2 horas
tramo_2_cnt = 0  # 2 a 5 horas
tramo_3_cnt = 0  # Más de 5 horas

for i in range(20):
    print(f"\n--- Vehículo {i+1} ---")
    patente = input("Ingrese patente del vehículo: ").strip().upper()
    while not patente:
        patente = input("Error: La patente es requerida: ").strip().upper()

    while True:
        try:
            horas = int(input(f"Cantidad de horas enteras para {patente}: "))
            if horas >= 0:
                break
            print("Error: Las horas no pueden ser negativas.")
        except ValueError:
            print("Error: Ingrese un número entero válido.")

    if horas <= 2:
        costo = 2000
        tramo_1_cnt += 1
    elif horas <= 5:
        costo = 3500
        tramo_2_cnt += 1
    else:
        costo = 5000
        tramo_3_cnt += 1

    total_recaudado += costo
    print(f"Vehículo {patente} | Cobro: ${costo:,} CLP")

print("\n--- Balance del Estacionamiento ---")
print(f"Total recaudado: ${total_recaudado:,} CLP")
print(f"Vehículos en Tramo 1 (hasta 2h): {tramo_1_cnt}")
print(f"Vehículos en Tramo 2 (2h a 5h): {tramo_2_cnt}")
print(f"Vehículos en Tramo 3 (más de 5h): {tramo_3_cnt}")
```

### Script 14: Clasificador censal de 40 personas

**Enunciado:** Desarrolla un script que solicite la edad de 40 personas. Para cada una, debe clasificarla en: Menor de edad, Mayor de edad, Adulto mayor, Cuarta edad en Chile. Al final, debe mostrar cuántas personas hay en cada categoría.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Recopilar de forma segura 40 edades consecutivas y clasificarlas según la demografía gerontológica de Chile.
- **Lógica de la solución:**
1. Definir cuatro contadores para almacenar las frecuencias de los intervalos etarios.
2. Correr bucle `for` con 40 iteraciones pidiendo la edad como entero en rango realista `[0, 120]`.
3. Comparar con condicionales anidados acumulando las estadísticas poblacionales al final.

```python
# Script 14: Clasificación demográfica (40 Personas)

print("--- Clasificador de Edades (40 Personas) ---")
menor_cnt = 0
mayor_cnt = 0
adulto_mayor_cnt = 0
cuarta_edad_cnt = 0

for i in range(40):
    while True:
        try:
            edad = int(input(f"Ingrese la edad de la persona {i+1} (0 a 120 años): "))
            if 0 <= edad <= 120:
                break
            print("Error: Ingrese una edad realista entre 0 y 120 años.")
        except ValueError:
            print("Error: Ingrese un número entero.")

    if edad < 18:
        menor_cnt += 1
    elif edad < 60:
        mayor_cnt += 1
    elif edad < 80:
        adulto_mayor_cnt += 1
    else:
        cuarta_edad_cnt += 1

print("\n--- Distribución Demográfica ---")
print(f"Menores de edad (0-17 años): {menor_cnt}")
print(f"Mayores de edad (18-59 años): {mayor_cnt}")
print(f"Adultos mayores (60-79 años): {adulto_mayor_cnt}")
print(f"Personas de la cuarta edad (>=80 años): {cuarta_edad_cnt}")
```

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Verificar la propiedad astronómica de bisiesto en 15 entradas del usuario de forma acumulada.
- **Lógica de la solución:**
1. Definir contadores `bisiestos = 0` y `no_bisiestos = 0`.
2. Iterar 15 veces solicitando años enteros válidos.
3. Aplicar la fórmula: divisible por 4 y no por 100, a menos que sea divisible por 400. Acumular y emitir resultados.

```python
# Script 15: Verificación de Años Bisiestos (15 registros)

print("--- Verificación de Años Bisiestos (15 registros) ---")
bisiestos = 0
no_bisiestos = 0

for i in range(15):
    while True:
        try:
            ano = int(input(f"Ingrese el año {i+1} a evaluar (mayor a 0): "))
            if ano > 0:
                break
            print("Error: El año debe ser mayor a cero.")
        except ValueError:
            print("Error: Ingrese un número entero.")

    es_bisiesto = (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0)
    if es_bisiesto:
        bisiestos += 1
        print(f"El año {ano} es BISIESTO.")
    else:
        no_bisiestos += 1
        print(f"El año {ano} es NO BISIESTO.")

print("\n--- Resumen de Registros ---")
print(f"Total años bisiestos: {bisiestos}")
print(f"Total años no bisiestos: {no_bisiestos}")
```

### Script 16: Paridad variable con cantidad n

**Enunciado:** Desarrolla un script que pida al usuario un número n. Luego, debe solicitar n números enteros e indicar para cada uno si es par o impar. Al final, debe mostrar: Cantidad de números pares e impares.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Generalizar el script de paridad permitiendo al estudiante decidir en tiempo de ejecución la cantidad de mediciones `n`.

1. Solicitar y validar que `n` sea un entero positivo.
2. Correr bucle `for` de rango `n` capturando enteros.
3. Evaluar con módulo `% 2` e incrementar estadísticas consolidando al final.

```python
# Script 16: Analizador de Paridad Variable (n números)

print("--- Analizador de Paridad Variable ---")
while True:
    try:
        n = int(input("¿Cuántos números enteros desea evaluar?: "))
        if n > 0:
            break
        print("Error: La cantidad de números debe ser mayor a cero.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

pares = 0
impares = 0

for i in range(n):
    while True:
        try:
            num = int(input(f"Ingrese el número entero {i+1} de {n}: "))
            break
        except ValueError:
            print("Error: Ingrese un entero válido.")

    if num % 2 == 0:
        pares += 1
        print(f"El número {num} es PAR.")
    else:
        impares += 1
        print(f"El número {num} es IMPAR.")

print("\n--- Resultados ---")
print(f"Números pares: {pares}")
print(f"Números impares: {impares}")
```

### Script 17: Caja registradora variable con n compras

**Enunciado:** Desarrolla un script que pida al usuario un número n, que representará la cantidad de compras a procesar. Para cada compra, debe pedir el monto total y aplicar estas reglas:
• Si la compra es mayor a $40.000 y hasta $100.000, se aplica un 10% de descuento
• Si la compra es superior a $100.000, se aplica un 20% de descuento
• Si la compra es de $40.000 o menos, no se aplica descuento
Al final, debe mostrar: Total vendido sin descuento, total descontado, total final recaudado.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Procesar un flujo variable `n` de ventas comerciales, determinando el descuento aplicable dinámicamente y la facturación acumulada.
- **Lógica de la solución:**
1. Validar el ingreso de `n` positivo.
2. Inicializar acumuladores de montos originales y montos descontados.
3. Recorrer `n` veces solicitando el costo base de la compra.
4. Clasificar el porcentaje de descuento, calcularlo, acumularlo en variables independientes y mostrar el reporte final.

```python
# Script 17: Procesamiento de n compras y descuentos

print("--- Procesamiento de n Compras ---")
while True:
    try:
        n = int(input("¿Cuántas compras desea procesar hoy?: "))
        if n > 0:
            break
        print("Error: La cantidad de compras debe ser mayor a cero.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

total_sin_descuento = 0.0
total_descontado = 0.0

for i in range(n):
    while True:
        try:
            monto = float(input(f"Monto de la compra {i+1} de {n}: $"))
            if monto >= 0:
                break
            print("Error: El monto no puede ser negativo.")
        except ValueError:
            print("Error: Ingrese un número válido.")

    if monto > 100000:
        descuento = monto * 0.20
    elif monto > 40000:
        descuento = monto * 0.10
    else:
        descuento = 0.0

    total_sin_descuento += monto
    total_descontado += descuento
    precio_final = monto - descuento
    print(f"Compra {i+1} | Original: ${monto:,.2f} | Descuento: ${descuento:,.2f} | Pagar: ${precio_final:,.2f}")

total_final = total_sin_descuento - total_descontado
print("\n--- Consolidado de Ventas ---")
print(f"Total vendido sin descuento: ${total_sin_descuento:,.2f} CLP")
print(f"Total descontado del día    : ${total_descontado:,.2f} CLP")
print(f"Total neto final recaudado  : ${total_final:,.2f} CLP")
```

### Script 18: Evaluación grupal de IMC para n personas

**Enunciado:** Desarrolla un script que pida al usuario un número n, correspondiente a la cantidad de personas a evaluar. Para cada persona, debe pedir: Nombre, Peso en kilogramos, Estatura en metros. Luego debe calcular el IMC = peso / (estatura²) y clasificar según la OMS: Bajo peso (<18.5), Normal (18.5-24.9), Sobrepeso (>=25). Al final, mostrar: Cantidad de personas en cada categoría y promedio de IMC del grupo.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Evaluar parámetros biométricos y estados de peso para un grupo variable `n` de personas, obteniendo la media ponderada de masa corporal.
- **Lógica de la solución:**
1. Capturar la longitud del grupo `n`.
2. Iterar `n` veces solicitando el nombre, peso (2 a 300 kg) y estatura (0.50 a 2.50 m) con validaciones anidadas.
3. Obtener el IMC individual, clasificar en contadores según rangos OMS y sumar al acumulador general para promediar al cierre.

```python
# Script 18: Sistema de Evaluación de IMC Variable

print("--- Sistema de Evaluación de IMC Variable ---")
while True:
    try:
        n = int(input("¿Cuántas personas se evaluarán en el grupo?: "))
        if n > 0:
            break
        print("Error: El número de personas debe ser mayor a cero.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

bajo_peso = 0
normal = 0
sobrepeso = 0
suma_imc = 0.0

for i in range(n):
    print(f"\n--- Persona {i+1} de {n} ---")
    nombre = input("Ingrese nombre de la persona: ").strip()
    while not nombre:
        nombre = input("El nombre es requerido: ").strip()

    while True:
        try:
            peso = float(input(f"Ingrese peso en kilogramos de {nombre} (2.0 a 300.0 kg): "))
            if 2.0 <= peso <= 300.0:
                break
            print("Error: Ingrese un peso realista.")
        except ValueError:
            print("Error: Ingrese un valor numérico decimal.")

    while True:
        try:
            estatura = float(input(f"Ingrese estatura en metros de {nombre} (0.50 a 2.50 m): "))
            if 0.50 <= estatura <= 2.50:
                break
            print("Error: Ingrese una estatura realista.")
        except ValueError:
            print("Error: Ingrese un valor numérico decimal.")

    imc = peso / (estatura ** 2)
    suma_imc += imc

    if imc < 18.5:
        bajo_peso += 1
        clasif = "Bajo peso"
    elif imc < 25.0:
        normal += 1
        clasif = "Normal"
    else:
        sobrepeso += 1
        clasif = "Sobrepeso"

    print(f"Persona: {nombre} | IMC: {imc:.2f} | Clasificación: {clasif}")

promedio_imc = suma_imc / n
print("\n--- Informe Grupal de Salud ---")
print(f"Cantidad bajo peso  : {bajo_peso}")
print(f"Cantidad peso normal: {normal}")
print(f"Cantidad sobrepeso  : {sobrepeso}")
print(f"Promedio de IMC del grupo: {promedio_imc:.2f}")
```

### Script 19: Calculadora aritmética controlada de n operaciones

**Enunciado:** Desarrolla un script que pida al usuario un número n, correspondiente a la cantidad de operaciones que desea realizar. Para cada operación, debe pedir: Primer número, Segundo número, Operación (+, -, *, /). El programa debe mostrar el resultado de cada operación. En caso de división, debe evitar dividir por cero. Al final, debe indicar cuántas operaciones de cada tipo se realizaron.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Simular una calculadora interactiva repetitiva para realizar `n` operaciones con protección matemática ante división por cero y reportar la frecuencia de operadores.
- **Lógica de la solución:**
1. Validar el entero `n`.
2. Inicializar cuatro contadores independientes de operadores.
3. Ejecutar el bucle de rango `n` solicitando ambos números y el operador string.
4. Evaluar con estructuras de decisión: realizar la operación e incrementar contadores, validando que el divisor sea diferente de cero antes de dividir.

```python
# Script 19: Calculadora repetitiva variable

print("--- Calculadora Repetitiva Variable ---")
while True:
    try:
        n = int(input("¿Cuántas operaciones matemáticas desea realizar?: "))
        if n > 0:
            break
        print("Error: El número de operaciones debe ser mayor a cero.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

sumas = 0
restas = 0
mults = 0
divs = 0

for i in range(n):
    print(f"\n--- Operación {i+1} de {n} ---")
    while True:
        try:
            num1 = float(input("Ingrese primer número: "))
            num2 = float(input("Ingrese segundo número: "))
            break
        except ValueError:
            print("Error: Ingrese valores numéricos.")

    while True:
        op = input("Seleccione operación (+, -, *, /): ").strip()
        if op in ["+", "-", "*", "/"]:
            break
        print("Error: Operador matemático inválido.")

    if op == "+":
        result = num1 + num2
        sumas += 1
        print(f"Resultado: {num1} + {num2} = {result}")
    elif op == "-":
        result = num1 - num2
        restas += 1
        print(f"Resultado: {num1} - {num2} = {result}")
    elif op == "*":
        result = num1 * num2
        mults += 1
        print(f"Resultado: {num1} * {num2} = {result}")
    else:
        if num2 == 0:
            print("Error crítico: División por cero no permitida en matemáticas reales.")
        else:
            result = num1 / num2
            divs += 1
            print(f"Resultado: {num1} / {num2} = {result}")

print("\n--- Estadísticas de Operaciones ---")
print(f"Sumas realizadas         : {sumas}")
print(f"Restas realizadas        : {restas}")
print(f"Multiplicaciones realizadas: {mults}")
print(f"Divisiones válidas realizadas: {divs}")
```

### Script 20: Cálculo de peso en n objetos celestes

**Enunciado:** Desarrolla un script que pida al usuario un número n, correspondiente a la cantidad de objetos a analizar. Para cada objeto, debe pedir: Masa en kilogramos y Cuerpo celeste (Tierra, Luna, Marte o Júpiter). El cálculo del peso debe realizarse con peso = masa * gravedad (Tierra: 9.8, Luna: 1.62, Marte: 3.71, Júpiter: 24.79). Mostrar el peso calculado y al final: promedio de pesos y cantidad de veces que se eligió cada cuerpo.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Evaluar y consolidar la fuerza de atracción gravitatoria en Newtons para `n` masas físicas en cuerpos del sistema solar.
- **Lógica de la solución:**
1. Capturar y validar el entero `n`.
2. Inicializar contadores por cada cuerpo celeste y acumulador general de peso.
3. Solicitar masa positiva y seleccionar el cuerpo. Multiplicar por la constante de gravedad del cuerpo seleccionado y mostrar el peso individual. Al cierre, reportar promedios globales.

```python
# Script 20: Calculadora gravitatoria variable

print("--- Calculadora Física Interplanetaria ---")
while True:
    try:
        n = int(input("¿Cuántos objetos desea evaluar?: "))
        if n > 0:
            break
        print("Error: La cantidad de objetos debe ser mayor a cero.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

tierra_cnt = 0
luna_cnt = 0
marte_cnt = 0
jupiter_cnt = 0
suma_pesos = 0.0

for i in range(n):
    print(f"\n--- Objeto {i+1} de {n} ---")
    while True:
        try:
            masa = float(input("Ingrese masa del objeto en kilogramos (mayor a 0): "))
            if masa > 0:
                break
            print("Error: La masa debe ser mayor a cero.")
        except ValueError:
            print("Error: Ingrese un valor numérico.")

    while True:
        cuerpo = input("Seleccione cuerpo celeste (Tierra / Luna / Marte / Júpiter): ").strip().capitalize()
        if cuerpo == "Tierra":
            g = 9.80
            tierra_cnt += 1
            break
        elif cuerpo == "Luna":
            g = 1.62
            luna_cnt += 1
            break
        elif cuerpo == "Marte":
            g = 3.71
            marte_cnt += 1
            break
        elif cuerpo in ["Júpiter", "Jupiter"]:
            g = 24.79
            jupiter_cnt += 1
            break
        else:
            print("Cuerpo celeste no válido. Reintente.")

    peso = masa * g
    suma_pesos += peso
    print(f"Masa: {masa} kg | Cuerpo: {cuerpo} (g = {g} m/s²) | Peso: {peso:.2f} N")

promedio_peso = suma_pesos / n
print("\n--- Reporte Interplanetario ---")
print(f"Promedio de pesos calculados: {promedio_peso:.2f} N")
print(f"Selecciones de la Tierra: {tierra_cnt}")
print(f"Selecciones de la Luna  : {luna_cnt}")
print(f"Selecciones de la Marte : {marte_cnt}")
print(f"Selecciones de la Júpiter: {jupiter_cnt}")
```
