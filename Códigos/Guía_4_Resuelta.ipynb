### Ejercicio 1: Cálculo de total de compra con funciones

**Enunciado:** Crear una función que reciba precio y cantidad y retorne el total. El programa debe pedir los datos al usuario y mostrar el resultado.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Modelar una función matemática básica que multiplique dos variables numéricas, abstrayendo la lógica del flujo de entrada.
- **Lógica de la solución:**
1. Definir la función `calcular_total(precio, cantidad)` que multiplica los parámetros y retorna el producto.
2. Capturar y validar en el programa principal que las entradas sean numéricas y no negativas.
3. Invocar la función con los parámetros y mostrar el resultado.

```python
def calcular_total(precio, cantidad):
    """Calcula el total a pagar basándose en el precio unitario y la cantidad."""
    return precio * cantidad

# Programa principal
print("--- Ejercicio 1: Cálculo de Compra ---")
while True:
    try:
        p = float(input("Ingrese el precio unitario del producto ($): "))
        c = int(input("Ingrese la cantidad de unidades a comprar: "))
        if p >= 0 and c >= 0:
            break
        print("Error: El precio y la cantidad no pueden ser negativos.")
    except ValueError:
        print("Error: Ingrese valores numéricos válidos.")

total = calcular_total(p, c)
print(f"Monto total final a pagar: ${total:,.2f} CLP")
```

### Ejercicio 2: Evaluación modular de temperatura

**Enunciado:** Crear una función que reciba una temperatura y retorne: 'Frío' si es menor a 12, 'Templado' entre 12 y 25, 'Calor' sobre 25.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Implementar una clasificación condicional modular de la temperatura ambiente utilizando rangos exclusivos.
- **Lógica de la solución:**
1. Definir `evaluar_temperatura(temperatura)`.
2. Evaluar mediante condicionales `if-elif-else` las fronteras exactas (menor a 12, entre 12 y 25, mayor a 25) y retornar el string resultante.
3. Capturar el valor real en el bloque principal y mostrar la clasificación.

```python
def evaluar_temperatura(temperatura):
    """Clasifica la temperatura en Frío, Templado o Calor."""
    if temperatura < 12:
        return "Frío"
    elif temperatura <= 25:
        return "Templado"
    else:
        return "Calor"

# Programa principal
print("--- Ejercicio 2: Clasificación de Temperatura ---")
while True:
    try:
        t = float(input("Ingrese la temperatura en grados Celsius (°C): "))
        break
    except ValueError:
        print("Error: Ingrese un valor numérico decimal.")

resultado = evaluar_temperatura(t)
print(f"La temperatura de {t}°C está clasificada como: {resultado}")
```

### Ejercicio 3: Contar números mayores a un valor de corte

**Enunciado:** Dada una lista de números, contar cuántos son mayores a un valor dado. Crear una función que reciba la lista y un número y retorne la cantidad de valores mayores.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Diseñar un algoritmo modular para filtrar y contar secuencialmente los elementos de una colección que superen un umbral.
- **Lógica de la solución:**
1. Definir `contar_mayores(lista, valor)`.
2. Inicializar un contador en cero.
3. Recorrer con un ciclo `for` los elementos de la lista y sumar 1 al contador si superan el valor.
4. Retornar el contador. El programa principal define una muestra y un límite y los evalúa.

```python
def contar_mayores(lista, valor):
    """Cuenta cuántos elementos de la lista superan el valor de corte dado."""
    contador = 0
    for num in lista:
        if num > valor:
            contador += 1
    return contador

# Programa principal
print("--- Ejercicio 3: Contar Mayores ---")
muestra = [10.5, 45.0, 78.2, 5.4, 99.1, 12.0, 33.3, 50.0, 88.8, 2.1]
limite = 35.0
cantidad = contar_mayores(muestra, limite)
print(f"Muestra analizada: {muestra}")
print(f"Valor límite de comparación: {limite}")
print(f"Cantidad de elementos mayores a {limite}: {cantidad}")
```

### Ejercicio 4: Cálculo de promedio de notas con validación

**Enunciado:** Crear una función que reciba una lista de notas y retorne el promedio. El programa debe pedir cuántas notas se ingresarán.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Promediar un conjunto variable de notas académicas chilenas válidas, encapsulando la suma y división aritmética.
- **Lógica de la solución:**
1. Definir `calcular_promedio(lista_notas)`.
2. Controlar la división por cero retornando `0.0` si la lista está vacía.
3. Recorrer la lista sumando sus elementos y retornar el resultado dividido por la longitud de la lista.
4. Capturar en el principal de forma interactiva e invocar la función.

```python
def calcular_promedio(lista_notas):
    """Calcula la media aritmética de una lista de notas. Evita divisiones por cero."""
    if len(lista_notas) == 0:
        return 0.0
    suma = 0.0
    for nota in lista_notas:
        suma += nota
    return float(suma / len(lista_notas))

# Programa principal
print("--- Ejercicio 4: Promedio de Notas ---")
while True:
    try:
        cant = int(input("¿Cuántas notas desea promediar hoy?: "))
        if cant > 0:
            break
        print("Error: Debe ingresar al menos una nota.")
    except ValueError:
        print("Error: Ingrese un entero válido.")

notas = []
for i in range(cant):
    while True:
        try:
            n = float(input(f"Ingrese la nota del estudiante {i+1} de {cant} (1.0 a 7.0): "))
            if 1.0 <= n <= 7.0:
                notas.append(n)
                break
            print("Error: Rango académico de notas en Chile es de 1.0 a 7.0.")
        except ValueError:
            print("Error: Ingrese una nota decimal válida.")

promedio = calcular_promedio(notas)
print(f"\nNotas registradas: {notas}")
print(f"Promedio final calculado: {promedio:.2f}")
```

### Ejercicio 5: Búsqueda secuencial de productos en inventario

**Enunciado:** Crear una función que reciba una lista y un elemento y retorne True si existe, False si no.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Simular una búsqueda secuencial de texto sin importar mayúsculas o minúsculas en una estructura de almacén.
- **Lógica de la solución:**
1. Definir `buscar_producto(lista_inventario, producto)`.
2. Recorrer la lista y comparar el producto en minúsculas.
3. Retornar `True` inmediatamente si hay coincidencia; si el ciclo termina sin hallarlo, retornar `False`.

```python
def buscar_producto(lista_inventario, producto):
    """Verifica si un ítem específico existe en el inventario."""
    for item in lista_inventario:
        if item.lower() == producto.lower():
            return True
    return False

# Programa principal
print("--- Ejercicio 5: Inventario ---")
inventario = ["arroz", "aceite", "tallarines", "azucar", "leche", "cafe"]
busqueda = input("Ingrese el producto a buscar en bodega: ").strip()
while not busqueda:
    busqueda = input("Error: La búsqueda no puede estar vacía. Reintente: ").strip()

existe = buscar_producto(inventario, busqueda)
if existe:
    print(f"El producto '{busqueda}' SÍ existe en bodega.")
else:
    print(f"El producto '{busqueda}' NO está registrado en bodega.")
```

### Ejercicio 6: Filtrado manual de valores mayores a 50

**Enunciado:** Crear una función que reciba una lista y retorne una nueva lista con valores mayores a 50.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Construir de forma limpia una sublista a partir de un listado numérico original, filtrando umbrales mayores a 50.
- **Lógica de la solución:**
1. Definir `filtrar_mayores_50(lista_original)`.
2. Crear una lista vacía `lista_filtrada`.
3. Iterar sobre la lista original agregando elementos mayores a 50. Retornar la nueva lista.

```python
def filtrar_mayores_50(lista_original):
    """Filtra y retorna una nueva lista con valores estrictamente superiores a 50."""
    lista_filtrada = []
    for val in lista_original:
        if val > 50:
            lista_filtrada.append(val)
    return lista_filtrada

# Programa principal
print("--- Ejercicio 6: Filtrado de Datos (> 50) ---")
datos = [12, 55, 78, 3, 99, 45, 50, 67, 2, 89]
filtrados = filtrar_mayores_50(datos)
print(f"Datos originales: {datos}")
print(f"Nueva lista filtrada: {filtrados}")
```

### Ejercicio 7: Registro y control financiero de ventas del día

**Enunciado:** Crear funciones que: permitan ingresar ventas (ciclo), calculen el total vendido, indiquen cuántas ventas superan los 10000.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Estructurar un sistema modular y transaccional mediante múltiples funciones que administren la carga y análisis del flujo de caja.
- **Lógica de la solución:**
1. Definir `ingresar_ventas()`: Capturar ventas en un bucle interactivo `while` hasta que el usuario digite 0 o negativo.
2. Definir `calcular_total_ventas(lista_ventas)`: Sumar montos recorriendo la lista.
3. Definir `contar_ventas_mayor_10000(lista_ventas)`: Contar elementos superiores a 10.000.

```python
def ingresar_ventas():
    """Registra ventas en un bucle interactivo hasta que el usuario decida finalizar."""
    ventas = []
    print("Ingrese los montos de ventas. Escriba '0' o un valor menor para finalizar.")
    while True:
        try:
            monto = float(input("Ingrese monto de venta: $"))
            if monto <= 0:
                break
            ventas.append(monto)
        except ValueError:
            print("Error: Ingrese un monto numérico válido.")
    return ventas

def calcular_total_ventas(lista_ventas):
    """Calcula la suma total acumulada de las ventas."""
    total = 0.0
    for v in lista_ventas:
        total += v
    return total

def contar_ventas_mayor_10000(lista_ventas):
    """Cuenta cuántas ventas del listado superan el valor límite de $10.000."""
    contador = 0
    for v in lista_ventas:
        if v > 10000:
            contador += 1
    return contador

# Programa principal
print("--- Ejercicio 7: Registro y Control de Ventas ---")
registro_ventas = ingresar_ventas()
total_vendido = calcular_total_ventas(registro_ventas)
ventas_top = contar_ventas_mayor_10000(registro_ventas)

print("\n--- Reporte Consolidado de Ventas ---")
print(f"Listado de ventas: {registro_ventas}")
print(f"Suma total vendida: ${total_vendido:,.2f} CLP")
print(f"Cantidad de ventas sobre $10.000: {ventas_top}")
```

### Ejercicio 8: Análisis térmico y límites de seguridad

**Enunciado:** Crear funciones que: generen o reciban una lista de temperaturas, calculen promedio, indiquen cuántas están fuera de rango (menor a 10 o mayor a 30).

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Evaluar parámetros térmicos en invernaderos de manera modular, consolidando la media e indicando alarmas de seguridad.
- **Lógica de la solución:**
1. Definir `calcular_promedio_temp(lista_temps)`: Retornar suma acumulada / longitud.
2. Definir `contar_fuera_rango(lista_temps)`: Bucle comparador incrementando el contador si el valor está fuera de `[10, 30]`. El principal simula el registro de mediciones.

```python
def calcular_promedio_temp(lista_temps):
    """Calcula la temperatura promedio diaria."""
    if len(lista_temps) == 0:
        return 0.0
    suma = 0.0
    for t in lista_temps:
        suma += t
    return float(suma / len(lista_temps))

def contar_fuera_rango(lista_temps):
    """Cuenta cuántas temperaturas registradas salen del rango de 10°C a 30°C."""
    contador = 0
    for t in lista_temps:
        if t < 10 or t > 30:
            contador += 1
    return contador

# Programa principal
print("--- Ejercicio 8: Análisis Térmico ---")
temps_dia = [11.5, 9.8, 25.4, 32.1, 8.5, 15.0, 22.3, 29.9, 30.5, 12.0]
promedio = calcular_promedio_temp(temps_dia)
fuera_rango = contar_fuera_rango(temps_dia)

print(f"Mediciones térmicas: {temps_dia}")
print(f"Temperatura promedio: {promedio:.1f}°C")
print(f"Mediciones fuera de rango de seguridad (10°C - 30°C): {fuera_rango}")
```

### Ejercicio 9: Gestión modular y registro de clientes

**Enunciado:** Crear funciones que: permitan agregar clientes a una estructura, mostrar todos los clientes, contar cuántos hay.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Desarrollar un CRUD modular básico para gestionar clientes de manera interactiva, controlando ingresos vacíos o duplicidades.
- **Lógica de la solución:**
1. Definir `agregar_cliente(lista_clientes, nombre_cliente)`: Retornar `True` tras insertar de forma segura.
2. Definir `mostrar_clientes(lista_clientes)`: Recorrer con `enumerate` para una salida bonita.
3. Definir `contar_clientes(lista_clientes)`: Retornar la longitud de la lista.

```python
def agregar_cliente(lista_clientes, nombre_cliente):
    """Añade un cliente a la lista si no está vacío ni duplicado."""
    if nombre_cliente and nombre_cliente not in lista_clientes:
        lista_clientes.append(nombre_cliente)
        return True
    return False

def mostrar_clientes(lista_clientes):
    """Muestra secuencialmente todos los clientes registrados."""
    print("\n--- Listado Oficial de Clientes ---")
    for idx, c in enumerate(lista_clientes):
        print(f" {idx+1}. {c}")

def contar_clientes(lista_clientes):
    """Retorna la cantidad total de clientes registrados."""
    return len(lista_clientes)

# Programa principal
print("--- Ejercicio 9: Gestión de Clientes ---")
base_clientes = []
print("Ingrese clientes. Escriba 'fin' para finalizar la carga.")
while True:
    cli = input("Nombre del cliente a registrar: ").strip()
    if cli.lower() == "fin":
        break
    if not cli:
        print("Error: El nombre del cliente no puede estar en blanco.")
        continue
    agregado = agregar_cliente(base_clientes, cli)
    if agregado:
        print(f"-> Cliente '{cli}' añadido exitosamente.")
    else:
        print("-> Error: Cliente ya se encuentra registrado.")

total_clientes = contar_clientes(base_clientes)
mostrar_clientes(base_clientes)
print(f"\nTotal general de clientes registrados: {total_clientes}")
```

### Ejercicio 10: Sistema integrador de pedidos y despacho logístico

**Enunciado:** Crear funciones que: permitan registrar pedidos (cliente + cantidad), calcular total de productos, identificar el cliente con mayor pedido.

#### Explicación de la Lógica (para estudiar):
- **Planteamiento del problema:** Simular un sistema logístico avanzado mediante colecciones compuestas (lista de diccionarios) con búsquedas manuales de máximos.
- **Lógica de la solución:**
1. Definir `registrar_pedido(lista_pedidos, cliente, cantidad)`: Insertar un diccionario conteniendo la información.
2. Definir `calcular_total_productos(lista_pedidos)`: Iterar sumando la clave `cantidad`.
3. Definir `obtener_cliente_max_pedido(lista_pedidos)`: Comparador manual de máximos para encontrar al cliente de mayor demanda.

```python
def registrar_pedido(lista_pedidos, cliente, cantidad):
    """Registra un nuevo pedido asociándolo a un cliente en la lista."""
    lista_pedidos.append({"cliente": cliente, "cantidad": cantidad})

def calcular_total_productos(lista_pedidos):
    """Calcula el total general de unidades solicitadas entre todos los pedidos."""
    total = 0
    for ped in lista_pedidos:
        total += ped["cantidad"]
    return total

def obtener_cliente_max_pedido(lista_pedidos):
    """Identifica de forma manual al cliente con el mayor volumen solicitado."""
    if len(lista_pedidos) == 0:
        return None
    pedido_max = lista_pedidos[0]
    for ped in lista_pedidos[1:]:
        if ped["cantidad"] > pedido_max["cantidad"]:
            pedido_max = ped
    return pedido_max

# Programa principal
print("--- Ejercicio 10: Portal Logístico de Pedidos ---")
pedidos_dia = []
print("Ingrese pedidos. Escriba 'fin' en el nombre del cliente para finalizar.")
while True:
    cli = input("Nombre del cliente: ").strip()
    if cli.lower() == "fin":
        break
    if not cli:
        print("Error: El nombre es requerido.")
        continue
    while True:
        try:
            cant = int(input(f"Cantidad de productos para {cli}: "))
            if cant > 0:
                break
            print("Error: La cantidad debe ser mayor a cero.")
        except ValueError:
            print("Error: Ingrese un entero válido.")
    registrar_pedido(pedidos_dia, cli, cant)
    print(f"-> Pedido guardado.")

total_unidades = calcular_total_productos(pedidos_dia)
pedido_estrella = obtener_cliente_max_pedido(pedidos_dia)

print("\n--- Reporte Consolidado de Despacho ---")
for ped in pedidos_dia:
    print(f"- {ped['cliente']}: {ped['cantidad']} unidades")
print(f"\nSuma global de unidades: {total_unidades}")
if pedido_estrella:
    print(f"Cliente estrella (Mayor pedido): {pedido_estrella['cliente']} con {pedido_estrella['cantidad']} unidades")
```
